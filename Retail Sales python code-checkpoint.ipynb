```python
import os
import random
from datetime import datetime, timedelta
from urllib.parse import quote_plus

import numpy as np
import pandas as pd
from sqlalchemy import create_engine
```

```python
# GENERATE MESSY SALES DATA
```

```python
# Set a fixed seed to ensure data reproducibility
np.random.seed(42)
random.seed(42)
```

```python
# List of products and their categories
products = {
    'Rice 5kg': 'Grocery', 'Wheat Flour 5kg': 'Grocery', 'Sugar 1kg': 'Grocery',
    'Cooking Oil 1L': 'Grocery', 'Tea Powder 250g': 'Grocery',
    'Soap': 'Personal Care', 'Shampoo': 'Personal Care', 'Toothpaste': 'Personal Care',
    'Biscuits': 'Snacks', 'Chips': 'Snacks', 'Namkeen 200g': 'Snacks',
    'Milk 1L': 'Dairy', 'Curd 500g': 'Dairy', 'Paneer 200g': 'Dairy',
    'Detergent 1kg': 'Household', 'Dishwash Liquid': 'Household'
}

# Standard price for each product
base_prices = {
    'Rice 5kg': 350, 'Wheat Flour 5kg': 220, 'Sugar 1kg': 45,
    'Cooking Oil 1L': 150, 'Tea Powder 250g': 120,
    'Soap': 35, 'Shampoo': 180, 'Toothpaste': 90,
    'Biscuits': 30, 'Chips': 20, 'Namkeen 200g': 60,
    'Milk 1L': 55, 'Curd 500g': 40, 'Paneer 200g': 80,
    'Detergent 1kg': 130, 'Dishwash Liquid': 95
}

customer_types = ['Regular', 'Walk-in', 'Wholesale']
```

```python
# Define the date range from January 2026 to June 2026
start_date = datetime(2026, 1, 1)
end_date = datetime(2026, 6, 30)
date_range_days = (end_date - start_date).days

NUM_RECORDS = 2800  # Total number of sales rows to create
```

```python
records = []

for i in range(NUM_RECORDS):
    product = random.choice(list(products.keys()))
    category = products[product]

    # Change the price slightly by adding or subtracting a small random number
    price = base_prices[product] + random.randint(-5, 5)
    quantity = random.randint(1, 20)

    # Choose a random date within our 6-month window
    random_days = random.randint(0, date_range_days)
    sale_date = start_date + timedelta(days=random_days)

    customer_type = random.choice(customer_types)

    records.append({
        'sale_id': i + 1,
        'product_name': product,
        'category': category,
        'quantity': quantity,
        'unit_price': price,
        'sale_date': sale_date.strftime('%Y-%m-%d'),
        'customer_type': customer_type
    })
```

```python
df = pd.DataFrame(records)
```

```python
# Introduce common data issues to simulate real-world data 
```

```python
# 1. Remove quantity values from 80 random rows

missing_qty_indices = np.random.choice(df.index, size=80, replace=False)
df.loc[missing_qty_indices, 'quantity'] = np.nan

```

```python
# 2. Remove price values from 50 random rows
missing_price_indices = np.random.choice(df.index, size=50, replace=False)
df.loc[missing_price_indices, 'unit_price'] = np.nan
```

```python
# 3. Copy 30 random rows and add them back to create duplicate data
duplicates = df.sample(30)
df = pd.concat([df, duplicates], ignore_index=True)
```

```python
# 4. Make 40 product names uppercase and add extra spaces
inconsistent_indices = np.random.choice(df.index, size=40, replace=False)
df.loc[inconsistent_indices, 'product_name'] = (
    df.loc[inconsistent_indices, 'product_name'].str.upper() + '  '
)
```

```python
# Save the data to a CSV file

df.to_csv('retail_sales_raw.csv', index=False)

print("Raw dataset created!")
print(f"Total records: {len(df)}")
print(df.head(10))
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# DATA CLEANING
```

```python
df = pd.read_csv('retail_sales_raw.csv')

print("\n=== Status before cleaning ===")
print(f"Total rows: {len(df)}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Missing quantity: {df['quantity'].isnull().sum()}")
print(f"Missing unit_price: {df['unit_price'].isnull().sum()}")
print(f"Unique product names (raw): {df['product_name'].nunique()}")
```

```python
# Remove Duplicates Rows

before = len(df)
df = df.drop_duplicates(subset=['sale_id'], keep='first')
after = len(df)
print(f"\nDuplicates removed: {before - after} rows")
print(f"Is sale_id unique now: {df['sale_id'].is_unique}")
```

```python
# Clean Product Name
# Remove extra spaces and fix the text capitalization (e.g., "SOAP  " becomes "Soap")

df['product_name'] = df['product_name'].str.strip().str.title()
print(f"Unique product names (after cleaning): {df['product_name'].nunique()}")
```

```python
# Fill missing quantity values 

# Fill empty quantity cells with the median quantity of that specific product
df['quantity'] = df.groupby('product_name')['quantity'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
# Fill missing unit_price values 
# Fill empty price cells with the average price of that specific product
df['unit_price'] = df.groupby('product_name')['unit_price'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
# Clean up numbers by converting quantity to integer and rounding the price
df['quantity'] = df['quantity'].round(0).astype(int)
df['unit_price'] = df['unit_price'].round(2)
```

```python
# Convert the date column to a real datetime type ----
# Change sale_date from text format to a proper date format
df['sale_date'] = pd.to_datetime(df['sale_date'])
```

```python
# Create new useful columns 
# Calculate the total sale amount and extract the month name and day name
df['total_amount'] = df['quantity'] * df['unit_price']
df['month'] = df['sale_date'].dt.month_name()
df['day_of_week'] = df['sale_date'].dt.day_name()
```

```python
# Final check 
print("\n=== Status after cleaning ===")
print(f"Total rows: {len(df)}")
print(f"Missing values:\n{df.isnull().sum()}")
print(f"\nData types:\n{df.dtypes}")
print("\nSample data:")
print(df.head())
```

```python
# Save the clean data to a new CSV file
df.to_csv('retail_sales_cleaned.csv', index=False)
print("\n'retail_sales_cleaned.csv' created!")
```

```python
# LOAD CLEAN DATA INTO MYSQL
```

```python
df = pd.read_csv('retail_sales_cleaned.csv')
```

```python
# Get the database password from system environment variables for security

db_password = os.environ.get('Your Pass')

if not db_password:
    raise ValueError(
        "Database password not found. "
        "Please set the DB_PASSWORD environment variable before running this script."
    )
```

```python
# Safely encode special characters in the password so the link works correctly
encoded_password = quote_plus(db_password)

DB_USER = 'root'
DB_HOST = 'localhost'
DB_NAME = 'retail_analytics'
```

```python
# Set up the connection link for the MySQL database
engine = create_engine(f'mysql+pymysql://{DB_USER}:{encoded_password}@{DB_HOST}/{DB_NAME}')
```

```python
# Upload the clean data into the 'sales' table
df.to_sql('sales', con=engine, if_exists='append', index=False)
print("Data successfully loaded into MySQL!")
```

```python
# Verify the upload by counting the total rows now in the database table
check = pd.read_sql('SELECT COUNT(*) as total_rows FROM sales', con=engine)
print(check)
```
